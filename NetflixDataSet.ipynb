# *Qualidade de Dados em Bases Públicas de Streaming: Diagnóstico Aplicado ao Dataset Netflix Movies and TV Shows*
O código desenvolvido visa apresentar a biblioteca pandas e suas funções para analise de dados, nesse caso com o objetivo de diagnosticar a qualidade dos dados do catálogo público da Netflix.



```python
import pandas as pd
df = pd.read_csv('netflix_titles.csv')
total_linhas, total_colunas = df.shape

# Mapeamento de valores e tipos
tipos = df.dtypes
nulos_absolutos = df.isnull().sum()
nulos_percentual = (nulos_absolutos / total_linhas * 100).round(1)

# Correção da coluna de data
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
nulos_data_antes = 10  # valor observado antes da conversão
datas_invalidas_depois = df['date_added'].isna().sum()

# Resultados
ranking_nulos = nulos_absolutos.sort_values(ascending=False)

print(f'Base: {total_linhas} linhas, {total_colunas} colunas')
print('\nTipos de dado por coluna:')
print(tipos)
print('\nValores ausentes por coluna (ordenado):')
print(ranking_nulos)
print(f'\nValores ausentes em date_added antes da conversão: {nulos_data_antes}')
print(f'Valores inválidos em date_added após conversão: {datas_invalidas_depois}')
```
